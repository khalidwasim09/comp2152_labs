```python
# monster.ipynb
from character import Character

class Monster(Character):
    def __init__(self):
        super().__init__()  # Call the parent constructor
        print("Monster created with combat strength:", self.combat_strength, "and health points:", self.health_points)

    # Destructor
    def __del__(self):
        super().__del__()  # Call the parent destructor
        print("The Monster object is being destroyed by the garbage collector")

    def monster_attacks(self):
        print("Monster attacks with combat strength:", self.combat_strength)
```
