```python
# hero.ipynb
from character import Character

class Hero(Character):
    def __init__(self):
        super().__init__()  # Call the parent constructor
        print("Hero created with combat strength:", self.combat_strength, "and health points:", self.health_points)

    # Destructor
    def __del__(self):
        super().__del__()  # Call the parent destructor
        print("The Hero object is being destroyed by the garbage collector")

    def hero_attacks(self):
        print("Hero attacks with combat strength:", self.combat_strength)
```
